```python
import mlflow

# Experiment lives in your user folder (MLflow experiments can't be created inside Git folders)
user = spark.sql("SELECT current_user()").first()[0]
mlflow.set_experiment(f"/Users/{user}/ed-forecaster")

with mlflow.start_run(run_name="tracking-smoke-test"):
    mlflow.log_param("purpose", "week 1 smoke test")
    mlflow.log_metric("dummy_mase", 1.0)
    mlflow.set_tag("week", "1")

print("Logged to experiment:", f"/Users/{user}/ed-forecaster")
```
